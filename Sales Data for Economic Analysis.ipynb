EXPLORATORY DATA ANALYSIS

```python
total_revenue = df['Revenue'].sum()
total_profit = df['Profit'].sum()
total_orders = df.shape[0]
avg_order_value = total_revenue / total_orders

print('Total Revenue:', total_revenue)
print("Total Profit", total_profit)
print("Total orders:", total_orders)
print("Average Order Value:", avg_order_value)
```

```python
category_summary = df.groupby('Product Category')[['Revenue', 'Profit']].sum()
category_summary
```

```python
country_summary = df.groupby('Country')[['Profit','Revenue']].sum()
country_summary
```

```python
gender_summary = df.groupby('Customer Gender')[['Revenue', 'Profit']].sum()
gender_summary
```

```python
bins = [0, 25, 35, 45, 55, 65, 100]
labels = ['Under 25', '25-34', '35-44', '45-54', '55-64', '65+']
df['Age Group'] = pd.cut(df['Customer Age'], bins = bins, labels = labels)

age_summary = df.groupby('Age Group')[['Revenue', 'Profit']].sum()
age_summary
```

```python
monthly_summary = df.groupby(['Year', 'Month'])[['Revenue', 'Profit']].sum()
monthly_summary
```

```python
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 
                'July', 'August', 'September', 'October', 'November',
                'December']
df['Month'] = pd.Categorical(df['Month'], categories = month_order, ordered = True)

monthly_summary = df.groupby(['Year', 'Month'])[['Revenue', 'Profit']].sum().reset_index()
monthly_summary
```

```python
monthly_summary['Period'] = monthly_summary['Year'].astype(str) + '-' + monthly_summary['Month'].astype(str)

plt.figure(figsize = (14,6))
plt.plot(monthly_summary['Period'], monthly_summary['Revenue'], label = 'Revenue', marker = 0)
plt.plot(monthly_summary['Period'], monthly_summary['Profit'], label = 'Profit', marker = 0)
plt.xticks(rotation = 45)
plt.title('Monthly Revenue and Profit (2015-2016)')
plt.xlabel('Month')
plt.ylabel('Amount ($)')
plt.legend()
plt.tight_layout()
plt.show
```

```python
monthly_summary = monthly_summary[monthly_summary['Revenue'] > 0]

plt.figure(figsize=(14,6))
plt.plot(monthly_summary['Period'], monthly_summary['Revenue'], label='Revenue', marker='o')
plt.plot(monthly_summary['Period'], monthly_summary['Profit'], label='Profit', marker='o')
plt.xticks(rotation=45)
plt.title('Monthly Revenue and Profit (2015-2016)')
plt.xlabel('Month')
plt.ylabel('Amount ($)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
category_summary = df. groupby('Product Category')[['Revenue', 'Profit']].sum()
category_summary['Profit Margin'] = category_summary['Profit'] / category_summary['Revenue']

fig, ax1 = plt.subplots(figsize = (8,5))

ax1.bar(category_summary.index, category_summary['Revenue'], color = 'steelblue')
ax1.set_ylabel('Revenue ($)', color = 'steelblue')
ax1.set_xlabel('Product Category')

ax2 = ax1.twinx()
ax2.plot(category_summary.index, category_summary['Profit Margin'], color = 'orange', marker = 0 , linewidth = 2)
ax2.set_ylabel('Profit Margin (%)', color = 'orange')

plt.title('Revenue vs Profit Margin by Product Category')
plt.tight_layout()
plt.show()
```

SUMMARY FINDINGS

- **Overall performance:** Total Revenue ~$22.3M, Total Profit ~$2.26M, overall profit margin ~10.1%
- **Product Category:** Bikes generate the most revenue but have the weakest margin (~2.9%); Accessories has the best margin (~18.6%) despite lower revenue
- **Country:** United States drives the most revenue but has one of the weakest margins (~6.8%); Germany produces less revenue but is by far the most profitable market (~22.6% margin)
- **Gender:** Revenue and margin are fairly similar between Female and Male customers
- **Age Group:** Customers aged 25-44 make up the majority of revenue; older age groups (55+) contribute comparatively little
- **Monthly Trend:** Revenue and profit both grew steadily from Jan 2015 to July 2016, with profit growing faster than revenue over time (improving margins). Dataset only covers Jan 2015 - July 2016.
